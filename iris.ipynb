```python
import pandas as pd
import numpy as np

# load iris dataset from project folder
df = pd.read_csv("iris.csv")

# numeric columns to perturb
num_cols = df.select_dtypes(include=np.number).columns.tolist()

# target size
target_n = 1500
current_n = len(df)
augmentation_n = target_n - current_n

rng = np.random.default_rng(42)

aug_rows = []
for _ in range(augmentation_n):
    row = df.sample(n=1, random_state=rng.integers(0, 1_000_000)).iloc[0].copy()
    for col in num_cols:
        noise = rng.normal(loc=0.0, scale=df[col].std() * 0.05)
        row[col] += noise
    aug_rows.append(row)

aug_df = pd.concat([df, pd.DataFrame(aug_rows)], ignore_index=True)
print("Original:", current_n, "Augmented:", len(aug_df))

aug_df.to_csv("iris_augmented.csv", index=False)
```
